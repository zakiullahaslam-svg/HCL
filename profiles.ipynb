```python
import pandas as pd

df = pd.read_csv("datasets/profiles.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['username'] = df['username'].fillna('unknown')
```

```python
df['gender'] = df['gender'].fillna('unknown')
df['birthday'] = df['birthday'].fillna('unknown')
```

```python
cols = ['joined','watching','completed','on_hold','dropped','plan_to_watch']
df[cols] = df[cols].fillna(0)  # Replace missing numeric activity counts with 0
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv("cleanedDatasets/profiles.csv", index=False)
```

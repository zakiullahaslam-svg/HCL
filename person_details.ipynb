```python
import pandas as pd

df = pd.read_csv("datasets/person_details.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['name'] = df['name'].fillna('unknown')
df['given_name'] = df['given_name'].fillna('unknown')
df['family_name'] = df['family_name'].fillna('unknown')
```

```python
df['birthday'] = df['birthday'].fillna('unknown')
```

```python
df['website_url'] = df['website_url'].fillna('N/A')
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv("cleanedDatasets/person_details.csv", index=False)
```

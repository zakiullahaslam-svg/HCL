```python
import pandas as pd

df = pd.read_csv("datasets/characters.csv")
```

```python
df.head()
```

```python
df = df.dropna(subset=['character_mal_id'])
```

```python
# Replace null 'about' with placeholder text
df['about'] = df['about'].fillna('No description available')
```

```python
df['favorites'] = df['favorites'].fillna(0).astype(int)
```

```python
df.shape
```

```python
# Clean text columns
text_cols = ['name', 'name_kanji', 'url', 'image']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv("cleanedDatasets/characters.csv", index=False)
```

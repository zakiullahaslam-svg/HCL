```python
import pandas as pd

df = pd.read_csv("datasets/details.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
# Remove empty or junk titles
df = df.dropna(subset=["title"])
df = df[~df["title"].str.fullmatch(r"[.\s]+")]
```

```python
df = df.drop_duplicates()
```

```python
# If title_japanese is NaN, copy the value from title
df["title_japanese"] = df["title_japanese"].fillna(df["title"])
```

```python
# Replace empty strings or NaNs with a placeholder like 'Unknown'
df['type'] = df['type'].replace('', 'Unknown')
df['type'] = df['type'].fillna('Unknown')
```

```python
# Fill missing numeric values with 0
df['score'] = df['score'].fillna(0)
df['scored_by'] = df['scored_by'].fillna(0)
df['episodes'] = df['episodes'].fillna(0)
df['rank'] = df['rank'].fillna(0)
df['rating'] = df['rating'].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
# Fill missing text values
df['synopsis'] = df['synopsis'].fillna("No synopsis")
df['start_date'] = df['start_date'].fillna("Unknown")
df['end_date'] = df['end_date'].fillna("Unknown")
df['season'] = df['season'].fillna("Unknown")
df['year'] = df['year'].fillna("Unknown")
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv("cleanedDatasets/details.csv", index=False)
```
